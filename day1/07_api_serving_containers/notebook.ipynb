## 🏛️ 1. The API Architecture

How does a mobile app actually talk to a Python Machine Learning model? 

Let's visualize the flow of data.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(12, 4))

# Draw the components
ax.add_patch(patches.Rectangle((0, 2), 2, 2, facecolor="#4A90D9"))
ax.text(1, 3, "Client\n(Web/Mobile)", color="white", ha="center", va="center", fontsize=12, fontweight="bold")

ax.add_patch(patches.Rectangle((4, 2), 2, 2, facecolor="#E94B4B"))
ax.text(5, 3, "FastAPI\n(REST Server)", color="white", ha="center", va="center", fontsize=12, fontweight="bold")

ax.add_patch(patches.Rectangle((8, 2), 2, 2, facecolor="#2ECC71"))
ax.text(9, 3, "ML Model\n(Scikit-Learn)", color="white", ha="center", va="center", fontsize=12, fontweight="bold")

# Draw the arrows
ax.annotate('', xy=(4, 3.5), xytext=(2, 3.5), arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10))
ax.text(3, 3.7, "JSON Request\n(Age: 35)", ha="center", va="bottom", fontsize=9)

ax.annotate('', xy=(8, 3.5), xytext=(6, 3.5), arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10))
ax.text(7, 3.7, "Validated\nPython Dict", ha="center", va="bottom", fontsize=9)

ax.annotate('', xy=(6, 2.5), xytext=(8, 2.5), arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10))
ax.text(7, 2.3, "Prediction: [1]", ha="center", va="top", fontsize=9)

ax.annotate('', xy=(2, 2.5), xytext=(4, 2.5), arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10))
ax.text(3, 2.3, "JSON Response\n{'default': true}", ha="center", va="top", fontsize=9)

ax.axis("off")
ax.set_xlim(-1, 11)
ax.set_ylim(0, 5)
plt.title("The Inference Lifecycle", fontsize=16)
plt.show()
```

## 🍰 2. The Docker Layer Cake

When you build a Docker Image, Docker executes your instructions step-by-step. Each step creates a **Layer**. 

If you change your python code, Docker doesn't want to re-install all your heavy libraries (like pandas and scikit-learn). It uses a cache! This is why the order of commands in a `Dockerfile` is so important.

```python
fig, ax = plt.subplots(figsize=(6, 8))

layers = [
    ("CMD ['uvicorn', ...]", "#95a5a6", "Runs the app (Changes often)"),
    ("COPY app/ /app", "#e74c3c", "Copies your python code (Changes often)"),
    ("RUN pip install -r reqs.txt", "#f1c40f", "Installs libraries (Changes rarely)"),
    ("COPY requirements.txt .", "#3498db", "Copies list of libraries (Changes rarely)"),
    ("FROM python:3.11-slim", "#2c3e50", "Base OS (Never changes)")
]

y_pos = 0
for title, color, desc in reversed(layers):
    ax.add_patch(patches.Rectangle((0, y_pos), 10, 1, facecolor=color, edgecolor="white", linewidth=2))
    ax.text(5, y_pos + 0.6, title, color="white", ha="center", va="center", fontsize=14, fontweight="bold")
    ax.text(5, y_pos + 0.2, desc, color="white", ha="center", va="center", fontsize=10)
    y_pos += 1

ax.annotate('', xy=(11, 4), xytext=(11, 1), arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10))
ax.text(11.5, 2.5, "Rebuilds start here\nif code changes!\n(Saves time!)", ha="left", va="center", fontsize=12)

ax.axis("off")
ax.set_xlim(-1, 15)
ax.set_ylim(0, 6)
plt.title("The Dockerfile Layer Cache Trick", fontsize=16, pad=20)
plt.show()
```

Now it is time to build it for real.

1. Open your code editor and look inside the `app/` folder.
2. Read `main.py` and `schemas.py`.
3. Open your terminal and run the API:

5. Stop the server (`Ctrl+C`).
6. Build the Docker image:
   `docker build -t my-ml-api .`
7. Run the Docker container:
   `docker run -p 8000:8000 my-ml-api`
