## Activity 2 — Explore the Parameter Growth Plot (Do this)

Instruction: Run the plotting cell below to visualise how model parameter counts grew over time. Observe the transformer inflection point around 2017.

```python
# ── Activity 2: Parameter growth plot (run) ─────────────────────────────────
param_data = [
    (2012, 'AlexNet',        60,       'DL'),
    (2014, 'VGG-16',         138,      'DL'),
    (2017, 'Transformer',    65,       'DL'),
    (2018, 'BERT-Base',      110,      'LLM'),
    (2018, 'GPT-1',          117,      'LLM'),
    (2019, 'GPT-2',          1500,     'LLM'),
    (2020, 'GPT-3',          175000,   'LLM'),
    (2021, 'BERT (large)',   340,      'LLM'),
    (2022, 'PaLM',           540000,   'LLM'),
    (2023, 'GPT-4*',         1000000,  'LLM'),
    (2024, 'Llama 3 (70B)',  70000,    'LLM'),
]
years = [d[0] for d in param_data]
names = [d[1] for d in param_data]
params = [d[2] for d in param_data]
cats = [d[3] for d in param_data]
cat_colors = {'DL':COLORS['dl'], 'LLM':COLORS['llm']}
fig, ax = plt.subplots(figsize=(12,5))
ax.bar(names, params, color=[cat_colors[c] for c in cats])
ax.set_yscale('log')
ax.set_ylabel('Parameters (log scale)')
ax.set_xticklabels(names, rotation=30, ha='right')
ax.set_title('Parameter Count Growth — Run and inspect')
plt.tight_layout()
plt.show()
```

## Activity 3 — Radar Chart Comparison (Optional, Do this)

Instruction: Run the cell below to render the capability radar chart comparing Classic ML, Deep Learning and LLMs. Use it to justify your answers in Activity 1.

```python
# ── Activity 3: Capability radar (run) ──────────────────────────────────────
categories = ['Inference Speed','Low Cost','Tabular Accuracy','Text/Image Accuracy','Interpretability','Low Data Need']
values_ml = [5,5,5,2,5,4]
values_dl = [3,3,3,5,2,2]
values_llm = [2,1,1,5,1,5]
# simple polar plot implementation
import math
N = len(categories)
angles = [n / float(N) * 2 * math.pi for n in range(N)]
angles += angles[:1]
vals_ml = values_ml + values_ml[:1]
vals_dl = values_dl + values_dl[:1]
vals_llm = values_llm + values_llm[:1]
fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, polar=True)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
ax.plot(angles, vals_ml, label='Classic ML')
ax.plot(angles, vals_dl, label='Deep Learning')
ax.plot(angles, vals_llm, label='LLM')
ax.fill(angles, vals_ml, alpha=0.1)
ax.fill(angles, vals_dl, alpha=0.1)
ax.fill(angles, vals_llm, alpha=0.1)
ax.legend(loc='upper right')
plt.show()
```
