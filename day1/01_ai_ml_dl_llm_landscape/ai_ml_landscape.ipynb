## 📐 Part 1 — The AI Hierarchy

Think of these fields as **Russian nesting dolls** (Matryoshka):

| Level | Field | Core Idea |
|-------|-------|-----------|
| 🌐 Outermost | **Artificial Intelligence** | Any technique that enables machines to mimic human intelligence |
| 📊 2nd layer | **Machine Learning** | Systems that *learn from data* instead of being explicitly programmed |
| 🧠 3rd layer | **Deep Learning** | ML using *multi-layer neural networks*; shines on images, audio, text |
| 💬 Innermost | **Large Language Models** | Deep Learning models trained on *trillions of tokens*; GPT, Gemini, LLaMA |

> **Key insight:** Everything that is ML is also AI, but not all AI is ML. Every LLM is Deep Learning, but most Deep Learning is not an LLM.

## 📊 Part 2 — ML Task Types

| Task Type | Learning Signal | Example Algorithms | Business Example |
|-----------|----------------|-------------------|------------------|
| **Supervised — Classification** | Labeled (categories) | Logistic Reg, XGBoost, RF | Fraud detection, Churn |
| **Supervised — Regression** | Labeled (numbers) | Linear Reg, GBM, SVR | Price forecast, Demand |
| **Unsupervised — Clustering** | No labels | K-Means, DBSCAN, GMM | Customer segmentation |
| **Unsupervised — Dim. Reduction** | No labels | PCA, UMAP, t-SNE | Data compression, Viz |
| **Reinforcement Learning** | Reward signal | DQN, PPO, A3C | Trading bot, Game AI |
| **Self-supervised / LLM** | Data is its own label | BERT, GPT, T5 | NLP, Code gen, Summarisation |

### 💬 LLM Adaptation Strategies

```
                       COST & TIME
  Low ◄─────────────────────────────────────► High
  │                                            │
  │  Prompt         Fine-tuning       RAG      │
  │  Engineering    (SFT/LoRA)        (Retrieval) │
  │                                            │
  ▼                                            ▼
 No model change   Weights updated   Knowledge added
 Zero labeled data Domain-specific   at inference time
```

**When to use which:**
- **Prompt Engineering** → quick prototype, zero labeled data
- **Fine-tuning** → domain-specific tone/format, ≥ 500 quality examples
- **RAG** → real-time knowledge, avoid hallucinations on enterprise documents

## 🎯 Part 3 — Model Capability Comparison

Different model types excel along different dimensions. In business we must choose deliberately:

| Dimension | Classic ML | Deep Learning | LLM |
|-----------|-----------|--------------|-----|
| **Speed (inference)** | ⚡⚡⚡ | ⚡⚡ | ⚡ |
| **Cost** | 💰 low | 💰💰 medium | 💰💰💰 high |
| **Accuracy (tabular)** | ⭐⭐⭐ | ⭐⭐ | ⭐ |
| **Accuracy (text/image)** | ⭐ | ⭐⭐⭐ | ⭐⭐⭐ |
| **Interpretability** | ⭐⭐⭐ | ⭐ | ⭐ |
| **Data Requirements** | Low–Medium | High | Extreme |

> **💡 In regulated industries (banking, insurance, healthcare):** Interpretability is non-negotiable. Classic ML with XAI often wins.

## 🤔 Part 4 — Which Technology Should I Use?

Use this decision framework for every new ML project:

```
START
  │
  ├─ Is the data structured/tabular? ──YES──► Classic ML (XGBoost, RF, LR)
  │                                          + XAI (SHAP) if regulated
  │
  └─ Is the data unstructured? ──────────────────────────────────────────
       │
       ├─ Images / Video? ──────► Deep Learning (CNN, ViT)
       │
       ├─ Audio / Speech? ───────► Deep Learning (Wav2Vec, Whisper)
       │
       └─ Text / Language? ──────────────────────────────────────────────
              │
              ├─ Need real-time knowledge? ──► RAG + LLM
              ├─ Need domain-specific format? ► Fine-tuned LLM
              └─ Zero labeled data? ──────────► Prompt Engineering + LLM
```

| 5 | **Interpretability matters** — in regulated industries, explainability is non-negotiable |
