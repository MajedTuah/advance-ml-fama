## 📝 Key Takeaways & Quiz

### Key Takeaways:
- **Global vs Local:** Global tells you how the model works generally. Local tells you why a specific decision was made.
- **SHAP vs LIME:** SHAP is more mathematically solid. LIME is often faster and can be easier to read for some audiences.
- **Translation:** Always translate technical SHAP values (+0.15) into plain-English business rules ("increased risk by 15%").
